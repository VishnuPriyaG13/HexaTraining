```python
product_data = [
    (1, "Laptop", 65000),
    (2, "Mobile", 30000),
    (3, "Tablet", 25000),
    (4, "Headphones", 5000)
]
```

```python
columns = ["product_id", "product_name", "price"]

product_df = spark.createDataFrame(product_data, columns)
```

```python
display(product_df)
```

```python
product_df.createOrReplaceTempView("products")
```

```python
avg_price_df = spark.sql("""
SELECT AVG(price) AS average_price
FROM products
""")
```

```python
display(avg_price_df)
```
