```python
account_data = [
    (1001, "Ravi", "Chennai", 50000),
    (1002, "Priya", "Coimbatore", 75000),
    (1003, "Arun", "Chennai", 30000),
    (1004, "Meena", "Madurai", 45000),
    (1005, "Kumar", "Coimbatore", 25000)
]
```

```python
columns = ["account_id", "customer_name", "city", "balance"]

bank_df = spark.createDataFrame(account_data, columns)
```

```python
display(bank_df)
```

```python
balance_report = bank_df.groupBy("city").sum("balance")
```
