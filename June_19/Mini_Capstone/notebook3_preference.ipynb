```python
preferences_path = "/Volumes/mini_cap/default/airline_data/Preferences.json"

df_preferences_raw = (
    spark.read
    .option("multiLine", "true")
    .json(preferences_path)
)

df_preferences_raw.printSchema()
```

```python
display(df_preferences_raw)
```

```python
from pyspark.sql.functions import col

df_preferences = df_preferences_raw.select(
    col("passenger_name"),
    col("meal"),
    col("seat"),
    col("extra_baggage")
)

df_preferences.printSchema()
display(df_preferences)
```
