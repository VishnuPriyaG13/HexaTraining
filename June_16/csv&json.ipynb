```python
doctor_visits = doctor_visits \
    .withColumn("tax",        round(col("bill_amount") * 0.05, 2)) \
    .withColumn("final_bill", round(col("bill_amount") + col("tax"), 2))

doctor_visits.select(
    "doctor_id", "doctor_name",
    "patient_name", "bill_amount",
    "tax", "final_bill"
).show()
```

```python
# calculate total revenue per doctor
revenue = doctor_visits.groupBy(
    "doctor_id", "doctor_name", "specialization", "city"
).agg(
    sum("bill_amount").alias("total_revenue"),
    count("visit_id").alias("total_visits")
)

# add ranking
window = Window.orderBy(col("total_revenue").desc())

doctor_ranked = revenue \
    .withColumn("revenue_rank",  rank().over(window)) \
    .withColumn("dense_rank",    dense_rank().over(window))

doctor_ranked.show()
```

```python
spec_summary = doctor_visits.groupBy("specialization") \
    .agg(
        count("doctor_id").alias("total_doctors"),
        count("visit_id").alias("total_visits"),
        sum("bill_amount").alias("total_revenue"),
        round(avg("bill_amount"), 2).alias("avg_bill"),
        round(avg("consultation_fee"), 2).alias("avg_consultation_fee")
    ) \
    .orderBy(col("total_revenue").desc())

spec_summary.show()
```

```python
silver_df = doctor_visits.select(
    "doctor_id", "doctor_name", "specialization",
    "city", "experience_years", "consultation_fee",
    "visit_id", "patient_name", "visit_date",
    "diagnosis", "bill_amount", "tax",
    "final_bill", "payment_status"
)
silver_df.write.mode("overwrite").saveAsTable("silver_doctor_visits")
```

```python
# gold report 1 - doctor performance
gold_doctor = spark.sql("""
    select doctor_id,
           doctor_name,
           specialization,
           city,
           count(visit_id)            as total_visits,
           sum(bill_amount)           as total_revenue,
           round(avg(bill_amount), 2) as avg_bill,
           rank() over (order by sum(bill_amount) desc) as revenue_rank
    from silver_doctor_visits
    group by doctor_id, doctor_name, specialization, city
    order by total_revenue desc
""")

gold_doctor.write.mode("overwrite").saveAsTable("gold_doctor_performance")
print("gold doctor report saved!")
gold_doctor.show()

# gold report 2 - specialization summary
gold_spec = spark.sql("""
    select specialization,
           count(distinct doctor_id)  as total_doctors,
           count(visit_id)            as total_visits,
           sum(bill_amount)           as total_revenue,
           round(avg(bill_amount), 2) as avg_bill
    from silver_doctor_visits
    group by specialization
    order by total_revenue desc
""")

gold_spec.write.mode("overwrite").saveAsTable("gold_specialization_summary")
print("gold specialization report saved!")
gold_spec.show()

# gold report 3 - city revenue summary
gold_city = spark.sql("""
    select city,
           count(distinct doctor_id)  as total_doctors,
           count(visit_id)            as total_visits,
           sum(bill_amount)           as total_revenue,
           round(avg(bill_amount), 2) as avg_bill
    from silver_doctor_visits
    group by city
    order by total_revenue desc
""")

gold_city.write.mode("overwrite").saveAsTable("gold_city_summary")
print("gold city report saved!")
gold_city.show()
```

```python
# final dashboard dataset combining all gold reports
# doctor performance
doctor_perf = spark.sql("""
    select d.doctor_id,
           d.doctor_name,
           d.specialization,
           d.city,
           d.experience_years,
           d.consultation_fee,
           count(v.visit_id)            as total_visits,
           sum(v.bill_amount)           as total_revenue,
           round(avg(v.bill_amount), 2) as avg_bill,
           sum(case when v.payment_status = 'Paid'    then v.bill_amount else 0 end) as paid_revenue,
           sum(case when v.payment_status = 'Pending' then v.bill_amount else 0 end) as pending_revenue,
           rank() over (order by sum(v.bill_amount) desc)                             as overall_rank,
           rank() over (partition by d.specialization order by sum(v.bill_amount) desc) as spec_rank,
           rank() over (partition by d.city          order by sum(v.bill_amount) desc) as city_rank
    from doctors d
    left join visits v on d.doctor_id = v.doctor_id
    group by d.doctor_id, d.doctor_name, d.specialization,
             d.city, d.experience_years, d.consultation_fee
    order by total_revenue desc
""")

doctor_perf.write.mode("overwrite").saveAsTable("gold_hospital_dashboard")
print("hospital operations analytics dashboard saved!")
doctor_perf.show()

# summary stats
print("\n--- dashboard summary ---")
spark.sql("""
    select count(doctor_id)             as total_doctors,
           sum(total_visits)            as total_visits,
           sum(total_revenue)           as total_revenue,
           round(avg(avg_bill), 2)      as overall_avg_bill,
           sum(paid_revenue)            as total_paid,
           sum(pending_revenue)         as total_pending
    from gold_hospital_dashboard
""").show()
```
