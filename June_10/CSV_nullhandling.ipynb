```python
df5=booking_df.fillna({"booking_amount":0})
df5.show()
```

```python
df6 = booking_df.withColumn(
    "data_quality_status",
    when(
        col("city").isNull() |
        col("provider").isNull() |
        col("booking_amount").isNull() |
        col("booking_status").isNull() |
        col("payment_mode").isNull(),
        "Incomplete"
    ).otherwise("Complete")
)
df6.show()
```

```python
df6.groupBy("data_quality_status").count().show()
```

```python
df6.filter(
    col("data_quality_status")=="Complete"
).show()
```

```python
df6.filter(
    col("data_quality_status")=="Incomplete"
).show()
```

```python
df7 = booking_df.fillna({"booking_amount":0})\
        .withColumn("tax", col("booking_amount")*0.05)

df7.show()
```

```python
df8 = df7.withColumn(
    "final_amount",
    col("booking_amount")+col("tax")
)
df8.show()
```

```python
revenue = booking_df.fillna({"booking_amount":0})\
            .filter(col("booking_status")=="Confirmed")\
            .agg(sum("booking_amount").alias("revenue"))

revenue.show()
```

```python
df9 = booking_df.fillna({"service_type":"Unknown"})
df9.groupBy("service_type").count().show()
```

```python
df10 = booking_df.fillna({"city":"Unknown"})
df10.groupBy("city").count().show()
```

```python
df11 = booking_df.fillna({"booking_amount":0})
df11.agg(avg("booking_amount").alias("avg_amount")).show()
```

```python
df12 = booking_df.dropna(subset=["booking_amount"])
df12.agg(avg("booking_amount").alias("avg_amount")).show()
```

```python
avg1 = booking_df.fillna({"booking_amount":0}) \
         .agg(avg("booking_amount").alias("avg1"))

avg2 = booking_df.dropna(subset=["booking_amount"]) \
         .agg(avg("booking_amount").alias("avg2"))

avg1.show()
avg2.show()
```
