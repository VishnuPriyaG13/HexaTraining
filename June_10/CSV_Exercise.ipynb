```python
from pyspark.sql.functions import *
```

```python
df=spark.read.csv('flightbookings.csv',header=True,inferSchema=True)
df.show()
```

```python
df.printSchema()
```

```python
df.select("passenger_name","from_city","to_city").show()
```

```python
conf_booking=df.filter(
    col("status")=="Confirmed"
)
conf_booking.show()
```

```python
can_booking=df.filter(
    col("status")=="Cancelled"
)
can_booking.show()
```

```python
pend_bookind=df.filter(
    col("status")=="Pending"
)
pend_bookind.show()
```

```python
ticket_price=df.filter(
    col("ticket_price")>20000
)
ticket_price.show()
```

```python
inter_flights=df.filter(
    col("to_city").isin("Dubai","Singapore","London")
)
inter_flights.show()
```

```python
df.count()
```

```python
by_airline=df.groupBy("airline").count()
by_airline.show()
```

```python
by_status=df.groupBy("status").count()
by_status.show()
```

```python
total_confirmed=df.filter(
    col("status")=="Confirmed")\
    .agg(sum("ticket_price").alias("total_revenue")
)
total_confirmed.show()
```

```python
avg_air=df.groupBy("airline")\
                   .agg(avg("ticket_price").alias("avg_price"))
avg_air.show()
```

```python
hig=df.agg(max("ticket_price").alias("highest_price"))
hig.show()
```

```python
low=df.agg(min("ticket_price").alias("lowest_price"))
low.show()
```

```python
df.orderBy(
    col("ticket_price").desc()
).show()
```

```python
df=df.withColumn(
    "tax",
    col("ticket_price") *0.5
)
df.show()
```

```python
df=df.withColumn(
    "final_price",
    col("ticket_price")+col("tax")
)
df.show()
```

```python
df=df.withColumn(
    "price_category",
    when(col("ticket_price")>=30000,"Premium")
    .when(col("ticket_price")>=10000,"Standard")
    .otherwise("Budget")
)
df.show()
```

```python
df.groupBy("price_category").count().show()
```

```python
df.filter(col("status")=="Confirmed")\
    .write.mode("overwrite")\
    .parquet("confirmed_flights.parquet")
```
