```python
from pyspark.sql.window import Window

df_win = df.withColumn(
    "total_amount",
    col("quantity") * col("price")
)
```

```python
window_spec = Window.orderBy(col("total_amount").desc())
df_win.withColumn(
    "row_num",
    row_number().over(window_spec)
).show()
```

```python
df_win.withColumn(
    "rank",
    rank().over(window_spec)
).show()
```

```python
df_win.withColumn(
    "dense_rank",
    dense_rank().over(window_spec)
).show()
```

```python
df_top5=df_win.withColumn(
    "rank",
    rank().over(window_spec)
)
df_top5.filter(col("rank")<=5).show()
```

```python
df_win.orderBy(col("total_amount").desc()).limit(1).show()
```

```python
city_window=Window.partitionBy("city")\
    .orderBy(col("total_amount").desc())
```

```python
df_win.withColumn(
    "rank",
    row_number().over(city_window)
).filter(col("rank") == 1).show()
```

```python
category_window = Window.partitionBy("category") \
                        .orderBy(col("total_amount").desc())

df_win.withColumn(
    "rank",
    row_number().over(category_window)
).filter(col("rank") == 1).show()
```

```python
running_window = Window.orderBy("order_id")

df_win.withColumn(
    "running_total",
    sum("total_amount").over(running_window)
).show()
```

```python
df_win.withColumn(
    "previous_revenue",
    lag("total_amount", 1).over(running_window)
).show()
```

```python
df_win.withColumn(
    "next_revenue",
    lead("total_amount", 1).over(running_window)
).show()
```

```python
df_win.withColumn(
    "previous_revenue",
    lag("total_amount").over(running_window)
).show()
```

```python
df_win.withColumn(
    "previous_revenue",
    lag("total_amount").over(running_window)
).withColumn(
    "revenue_difference",
    col("total_amount") - col("previous_revenue")
).show()
```

```python
city_window = Window.partitionBy("city") \
                    .orderBy(col("total_amount").desc())

df_win.withColumn(
    "rank",
    row_number().over(city_window)
).filter(col("rank") <= 2).show()
```

```python
category_window = Window.partitionBy("category") \
                        .orderBy(col("total_amount").desc())

df_win.withColumn(
    "rank",
    row_number().over(category_window)
).filter(col("rank") <= 3).show()
```

```python
df_win.withColumn(
    "rank",
    dense_rank().over(window_spec)
).select(
    "order_id",
    "product",
    "city",
    "total_amount",
    "rank"
).show()
```

```python
#### Min ETL
df_etl = df.withColumn(
    "total_amount",
    col("quantity") * col("price")
).withColumn(
    "tax",
    col("total_amount") * 0.05
).withColumn(
    "final_amount",
    col("total_amount") + col("tax")
)

df_etl.show()
```

```python
completed_df = df_etl.filter(
    col("status") == "Completed"
)

completed_df.show()
```

```python
completed_df.select(
    sum("total_amount").alias("total_revenue")
).show()
```

```python
city_report = completed_df.groupBy("city") \
                          .agg(sum("total_amount")
                          .alias("city_revenue"))

city_report.show()
```

```python
category_report = completed_df.groupBy("category") \
                              .agg(sum("total_amount")
                              .alias("category_revenue"))

category_report.show()
```
