```python
patients_df = spark.read.csv(
    "patients.csv",
    header=True,
    inferSchema=True
)

appointments_df = spark.read.csv(
    "appointments.csv",
    header=True,
    inferSchema=True
)
```

```python
preferences_df = spark.read \
    .option("multiline", "true") \
    .json("patient_preferences.json")
```

```python
patients_df = patients_df.fillna({
    "city": "Unknown",
    "blood_group": "Not Available"
})

appointments_df = appointments_df.fillna({
    "consult": 0
})
```

```python
final_df = patients_df.join(
    appointments_df,
    "patient_id",
    "left"
).join(
    preferences_df,
    "patient_id",
    "left"
)

final_df.show()
```

```python
final_df = final_df.withColumn(
    "age_group",
    when(col("age") < 30, "Young")
    .when((col("age") >= 30) & (col("age") < 50), "Middle Age")
    .otherwise("Senior")
)

final_df.show()
```

```python
revenue_metrics = final_df.groupBy("department").agg(
    sum("consult").alias("department_revenue")
)

revenue_metrics.show()
```

```python
patient_spending = final_df.groupBy(
    "patient_id",
    "patient_name"
).agg(
    sum("consult").alias("total_spending")
)

patient_spending.show()
```

```python
department_revenue = final_df.groupBy(
    "department"
).agg(
    sum("consult").alias("total_revenue")
)

department_revenue.show()
```

```python
final_df.write.mode("overwrite").parquet(
    "hospital_analytics_output"
)
```

```python
print("===== HOSPITAL ANALYTICS REPORT =====")

print("Total Patients:",
      patients_df.count())

print("Total Appointments:",
      appointments_df.count())

patient_spending.show()

department_revenue.show()

revenue_metrics.orderBy(
    col("department_revenue").desc()
).show()
```
