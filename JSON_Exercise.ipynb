```python
from pyspark.sql import SparkSession
spark=SparkSession.builder \
.appName('RDD')\
.getOrCreate()
```

```python
from google.colab import files
uploaded=files.upload
```

```python
hotels_df = spark.read.option(
"multiline",
"true"
).json("hotels.json")
hotels_df.show(truncate=False)
hotels_df.printSchema()
```

```python
from pyspark.sql.functions import *
```

```python
hotels_df.show(truncate=False)
```

```python
hotels_df.select("hotel_name","city","rating").show()
```

```python
lux=hotels_df.filter(
    col("category")=="Luxury"
)
lux.show()
```

```python
rating=hotels_df.filter(
    col("rating")>4.5
)
rating.show()
```

```python
rooms=hotels_df.filter(
    col("rooms_available")>15
)
rooms.show()
```

```python
price_df=hotels_df.filter(
    col("price_per_night")>10000
)
price_df.show()
```

```python
dl=hotels_df.filter(
    col("city").isin("Dubai","London")
)
dl.show()
```

```python
con=hotels_df.filter(
    col("contact.phone").isNull()
)
con.show()
```

```python
con=hotels_df.filter(
    col("contact.email").isNull()
)
con.show()
```

```python
hotels_df.select(
    "hotel_name",
    "contact.phone",
    "contact.email"
).show()
```

```python
wifi=hotels_df.filter(array_contains(
    col("amenities"),"wifi")
)
wifi.show()
```

```python
wifi=hotels_df.filter(array_contains(
    col("amenities"),"spa")
)
wifi.show()
```

```python
hotels_df.groupBy("city").count().show()
```

```python
hotels_df.groupBy("category").count().show()
```

```python
avg_rating=hotels_df.groupBy("category").agg(
    avg("rating").alias("avg_rating")
)
avg_rating.show()
```

```python
avg_price=hotels_df.groupBy("city").agg(
    avg("price_per_night").alias("avg_price")
)
avg_price.show()
```

```python
hotels_df.agg(max("rating")).show()
```

```python
hotels_df = hotels_df.withColumn(
    "total_potential_revenue",
    col("rooms_available") * col("price_per_night")
)
hotels_df.show()
```

```python
hotels_df.orderBy(col("rating").desc()).show()
```

```python
flattened_df = hotels_df.select(
    "hotel_id",
    "hotel_name",
    "city",
    "category",
    "rating",
    "rooms_available",
    "price_per_night",
    "contact.phone",
    "contact.email"
)

flattened_df.write.mode("overwrite").parquet("hotels_flattened.parquet")
```
