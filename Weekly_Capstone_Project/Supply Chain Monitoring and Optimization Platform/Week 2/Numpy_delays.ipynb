```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/orders_01.csv')
display(df)
```

```python
df = df.dropna(subset=['supplier_id'])
median_qty = df['quantity'].median()
df['quantity'] = df['quantity'].fillna(median_qty)
df['order_date'] = pd.to_datetime(df['order_date'], format='mixed')
df['delivery_date'] = pd.to_datetime(df['delivery_date'], format='mixed')
df['supplier_id'] = df['supplier_id'].astype(int)
df['quantity'] = df['quantity'].astype(int)
```

```python
today = pd.Timestamp.today().normalize()
```

```python
df['delay_days'] = np.where(
    df['delivery_date'].notna(),
    (today - df['delivery_date']).dt.days,
    (today - df['order_date']).dt.days
)
```

```python
df['is_delayed'] = np.where(df['delay_days'] > 0, 1, 0)
```

```python
print("--- Final Processed Data ---")
print(df[['order_id', 'supplier_id', 'product_id', 'quantity',
          'order_date', 'delivery_date', 'delay_days', 'is_delayed']])

print("\n--- Delay Summary ---")
print(f"Total orders: {len(df)}")
print(f"Delayed orders: {df['is_delayed'].sum()}")
print(f"On-time orders: {len(df) - df['is_delayed'].sum()}")

print("\n--- Delayed Orders Only ---")
print(df[df['is_delayed'] == 1][['order_id', 'supplier_id', 'delay_days']])
```
