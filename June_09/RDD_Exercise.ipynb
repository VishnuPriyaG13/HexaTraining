```python
courses = [
"Python SQL Spark",
"Azure Databricks Spark",
"Python Pandas NumPy",
"SQL Spark DeltaLake",
"Azure Python SQL"
]
```

```python
course_rdd = spark.sparkContext.parallelize(courses)
```

```python
words_rdd = course_rdd.flatMap(lambda x: x.split())
print(words_rdd.collect())
```

```python
print(words_rdd.count())
```

```python
print(words_rdd.distinct().collect())
```

```python
sort=words_rdd.sortBy(lambda x:x)
print(sort.collect())
```

```python
freq = words_rdd.countByValue()
for word, count in freq.items():
    print(word, ":", count)
```

```python
res=words_rdd.filter(lambda x: "Python" in x)
print(res.collect())
```

```python
res=words_rdd.filter(lambda x: "Spark" in x)
print(res.collect())
```

```python
res=words_rdd.map(lambda x: x.upper())
print(res.collect())
```

```python
longest = words_rdd.reduce(
    lambda x, y: x if len(x) > len(y) else y
)
print(longest)
```

```python
shortest = words_rdd.reduce(
    lambda x, y: x if len(x) < len(y) else y
)
print(shortest)
```
